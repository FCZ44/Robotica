### **Desempeño cinemático**

Los índices cinemáticos dependen únicamente de los parámetros geométricos y de movimiento de un robot, sin considerar las fuerzas y torques que actúan sobre el sistema ni sus propiedades de masa e inercia. Por lo tanto, las métricas cinemáticas incluyen límites de movimiento, como rango de movimiento, evitación de colisiones, velocidad máxima y aceleración máxima, así como indicadores de precisión, como exactitud, repetibilidad y resolución.

### **Desempeño estático y dinámico**

Las métricas estáticas evalúan el desempeño de un robot bajo el efecto de una carga y son comunes en operaciones relativamente lentas (con aceleraciones limitadas). Las métricas dinámicas generalizan las formulaciones estáticas al considerar los efectos inerciales. Los indicadores de desempeño estático y dinámico más comunes están vinculados a la transmisión de fuerzas, la rigidez y el consumo de energía. Estas métricas son ampliamente utilizadas y resultan críticas para robots en entornos dinámicos, como robots submarinos y aéreos, o aquellos que requieren equilibrio dinámico para funcionar, como los bípedos.

**1. Desacoplamiento de la Muñeca ($P_3$)**
Calculamos las coordenadas de la articulación 3 restando la contribución del último eslabón $L_3$:

<div align = center>

${}^0x_3 = x_P - {}^{0}x_{1} - L_3 \cos(\theta_P)$

${}^0y_3 = y_P - {}^{0}y_{1} - L_3 \sin(\theta_P)$

</div>

**2. Resolución de las Articulaciones 1 y 2 (Hombro y Codo)**
Trazando un vector radial $r$ desde el origen de la base hasta la muñeca, obtenemos $r = \sqrt{({}^0x_3)^2 + ({}^0y_3)^2}$. Aplicando trigonometría sobre el triángulo formado por $L_1$, $L_2$ y $r$:

* Ángulo del Codo (${}^{1}\theta_2$):

<div align = center>

  $\beta = \cos^{-1}\left(\frac{L_1^2 + L_2^2 - r^2}{2 L_1 L_2}\right)$

  ${}^{1}\theta_2 = \pi - \beta$ *(Considerando la configuración de codo abajo/arriba según el signo)*

</div>

* Ángulo de la Base (${}^{0}\theta_1$):

<div align = center>

  $\alpha = \cos^{-1}\left(\frac{L_1^2 + r^2 - L_2^2}{2 L_1 r}\right)$
  
  $\psi = \text{atan2}({}^0y_3, {}^0x_3)$

  ${}^{0}\theta_1 = \psi \pm \alpha$

</div>

**3. Orientación Final (${}^{2}\theta_3$)**
Finalmente, como la orientación en el plano es la suma algebraica de los giros, despejamos la tercera articulación:

<div align = center>

${}^{2}\theta_3 = \theta_P - {}^{0}\theta_1 - {}^{1}\theta_2$

</div>

#### Resultado Final: Modelo Cinemático Inverso Analítico

Para cualquier punto objetivo definido por el vector de postura ${}^0\xi_P = [x_P, y_P, \theta_P]^T$, la configuración de las articulaciones $q = [{}^0\theta_1, {}^1\theta_2, {}^2\theta_3]^T$ se obtiene mediante las siguientes ecuaciones cerradas:

$$q = \begin{pmatrix}
{}^0\theta_1 \\
{}^1\theta_2 \\
{}^2\theta_3
\end{pmatrix} =
\begin{pmatrix}
\text{atan2}({}^0y_3, {}^0x_3) \pm \cos^{-1}\left(\frac{L_1^2 + r^2 - L_2^2}{2 L_1 r}\right) \\
\pi - \cos^{-1}\left(\frac{L_1^2 + L_2^2 - r^2}{2 L_1 L_2}\right) \\
\theta_P - {}^0\theta_1 - {}^1\theta_2
\end{pmatrix}$$


#### Variables intermedias de cálculo

Para resolver el sistema anterior, es necesario calcular previamente los siguientes parámetros geométricos:

* Coordenadas de la muñeca ($P_3$):

<div align = center>

${}^0x_3 = x_P - {}^0x_1 - L_3 \cos(\theta_P)$

${}^0y_3 = y_P - {}^0y_1 - L_3 \sin(\theta_P)$

</div>

* Radio efectivo ($r$): Distancia euclidiana desde la base hasta la articulación 3.

<div align = center>

$r = \sqrt{({}^0x_3)^2 + ({}^0y_3)^2}$

</div>

Nota: La solución para ${}^0\theta_1$ presenta un signo $\pm$, lo que indica la existencia de dos configuraciones posibles (codo arriba y codo abajo) para alcanzar el mismo punto en el espacio de trabajo. El robot puede alcanzar todos los puntos donde la cinemática inversa tiene solución real, es decir, donde el radio efectivo r hasta la muñeca $P_3$ satisface la desigualdad triangular.

<div align = center>

$-1 \leq \frac{L_1^2 + L_2^2 - r^2}{2 \cdot L_1 \cdot L_2} \leq 1$

</div>

Donde:

<div align = center>

$r = \sqrt{({}^0x_3)^2 + ({}^0y_3)^2}$

</div>

que es el radio efectivo hasta la muñeca desacoplada, lo que se simplifica a:

<div align = center>

$|L_1 - L_2| \leq r \leq L_1 + L_2$

</div>

donde r es es la distancia euclidiana desde la base hasta la articulación 3 desacoplada.

Para implementar esto en nuestro paquete de trabajo de ROS2, se hace un cambio al script base `scara_control.py` proporcionado por el profesor, el script con la generación de la trayectoria es ahora `scara_control_2.py`. El script básico (`scara_control.py`) publica únicamente dos posiciones dadas de forma alternada usando `time.sleep()` para esperar entre ellas. No interpola puntos intermedios, por lo que el efector final sigue una trayectoria curva en el espacio cartesiano, ya que el controlador de Gazebo interpola directamente en espacio de juntas.

El script `scara_control_2.py` tiene ahora las siguientes características:

  - *Interpolación carteasiana lineal:* En lugar de saltar de ángulo a ángulo, se generan puntos intermedios (`PUNTOS_POR_SEGMENTO`) distribuidos uniformemente a lo largo de la línea recta que une P1 y P2 en el espacio cartesiano (x, y, $\theta$). En cada punto se resuelve la cinemática inversa, garantizando que el efector siga una trayectoria recta visible en Gazebo y RViz.

  - *Cinemática iversa en tiempo de configuración:* En el script básico, cin_inv() está definida pero nunca se llama. Para el script `scara_control_2.py` cin_inv() se invoca para cada uno de los puntos interpolados, calculando ($\theta_1, \theta_2, \theta_3$) a partir de la postura cartesiana (x, y, $\theta$) del efector final.

  - *Unwrap incremental de $\theta_1$:* acos() siempre devuelve valores en [0, π], lo que provoca saltos discontinuos en $\theta_1$ cuando el robot cruza ciertas configuraciones. `unwrap_incremental()` detecta y corrige esos saltos comparando cada nuevo valor con el anterior y tomando el camino más corto ($\Delta\theta$ en [$-\pi, \pi$]), manteniendo la continuidad del movimiento.

  - *Pre-cálculo de toda la trayectoria:* `construir_trayectoria()` genera la lista completa de ángulos al importar el módulo y antes de que el nodo arranque. El callback del timer solo recorre esa lista punto a punto sin cálculos en tiempo real, lo que hace un control más predecible y eficiente.

  - *Control por timer sin `time.sleep()`:* El script básico (`scara_control.py`) usa `time.sleep()` dentro del callback, lo que bloquea el hilo del executor de ROS2 e impide procesar otros eventos. Aquí el timer dispara cada `PERIODO_TIMER` segundos y avanza un índice, respetando el modelo asíncrono de rclpy.

El script `scara_control_2.py` se muestra en el repositorio. Para la implementación de este script, se hace el cambio en el `setup.py` y se realiza `colcon_build` después `source install/setup.bash`.

#### **Índice de manupulabilidad de Yoshikawa**

Para un robot planar 3R, el índice de manipulabilidad se calcula como:

<div align = "center">

$\mu = \sqrt{det(J * J^T)}$

</div>

Y para este robot en específico, un robot SCARA de 3 grados de libertad, el índice de manipulabilidad se puede reducir como se muestra a continuación:

<div align = "center">

$\mu = |L_1 L_2 sin(\theta_2)| L_3$

</div>

Se procede a graficar el índice de manipulabilidad, según el resultado que el script `indiceManipulabilidad.m` nos da:

<div align = "center">

![Indice de manipulabilidad](imagenes/indiceManipulabilidad.png)

</div>

En general, cuando $\mu$ se acerca a cero el robot está cerca de una singularidad (puede perder la capacidad de movimiento y el jacobiano inverso se vuelve inestable); con un $\mu$ alto, el robot tiene mayor "destreza" y se puede mover más fácilmente.

En la gráfica podemos ver que $\mu$ no llega a cero en ningún momento, sólo en los extremos que es donde el robot está extendido lateralmente, son los puntos más cercanos a singularidades. Cuando el robot se encuentra en el tiempo = 3 ó tiempo = 7, el robot tiene el mayor valor de manipulabilidad. Cuando el robot se encuentra en el tiempo = 5, el robot se encuentra en la mitad del recorrido y es donde sus articulaciones están lo más "retraídas" haciendo que $\mu$ caiga localmente, esto no es una singularidad pero es donde el robot tiene menor destreza durante el recorrido de la trayectoria.

#### **Potencia mecánica requerida por cada motor y potencia total para el robot.**


Una vez obtenidas las trayectorias articulares, sus velocidades y aceleraciones, se calcularon los torques requeridos por cada motor utilizando el modelo dinámico del robot.

Para cada instante de tiempo se utilizan:

* **Posiciones articulares:**
  $$\theta_1, \theta_2, \theta_3$$
* **Velocidades articulares:**
  $$\dot{\theta}_1, \dot{\theta}_2, \dot{\theta}_3$$
* **Aceleraciones articulares:**
  $$\ddot{\theta}_1, \ddot{\theta}_2, \ddot{\theta}_3$$

El código implementa las ecuaciones dinámicas obtenidas mediante el método de Euler-Lagrange. Los términos auxiliares denominados `sigma` son utilizadas para simplificar el cálculo de los torques.

Como resultado se obtienen los torques requeridos en cada articulación:

$$\tau_1, \tau_2, \tau_3$$

Los cuales representan el esfuerzo mecánico que debe generar cada motor para seguir la trayectoria planeada.

##### Potencia mecánica

La potencia mecánica instantánea requerida por cada motor se calcula de la siguiente forma:

$$P_i = \tau_i \dot{\theta}_i$$

Donde:
* **$P_i$:** Potencia mecánica de la articulación $i$ [W].
* **$\tau_i$:** Torque de la articulación $i$ [N·m].
* **$\dot{\theta}_i$:** Velocidad angular de la articulación $i$ [rad/s].

En la implementación se utiliza el valor absoluto para obtener únicamente la magnitud de la potencia requerida:

$$P_i = |\tau_i \dot{\theta}_i|$$

La potencia total demandada por el robot se obtiene sumando las potencias de las tres articulaciones:

$$P_T = P_1 + P_2 + P_3$$

Este último valor permite estimar la capacidad mínima que debe tener todo el sistema de accionamiento para ejecutar la trayectoria especificada.

Los resultados se representan mediante dos gráficas, optenidas con el script `TyPEspacioTrabajo`. La primera muestra la evolución temporal de los torques requeridos en cada articulación, mientras que la segunda presenta la potencia mecánica desarrollada por cada motor y la potencia total requerida durante la ejecución de la trayectoria planificada.

Se observa que la segunda articulación es la que requiere el mayor esfuerzo dinámico, alcanzando valores cercanos a -4.6 N·m. Esto indica que el motor asociado a dicha articulación se debe seleccionar con una capacidad de torque superior a la de las demás juntas, o en su caso, si todos los motores son iguales, entonces la restricción de mínimo torque se determina por la segunda articulación.

La primera articulación presenta un torque máximo aproximado de 2.3 N·m, mientras que la tercera articulación requiere esfuerzos considerablemente menores, con magnitudes inferiores a 1 N·m.

Asimismo, los torques convergen a valores cercanos a cero al inicio y al final de la trayectoria, este comportamiento es el esperado debido a las condiciones de reposo impuestas por el polinomio de quinto orden utilizado en la planeación.

La segunda articulación es la principal responsable de este incremento, ya que concentra los mayores esfuerzos dinámicos del sistema. Por su parte, la tercera articulación presenta una contribución reducida al consumo total de potencia. Coincide que la segunda articulación presenta más torque y también más potencia.

Nótese que la potencia mecánica es prácticamente nula al inicio y al final del movimiento. Esto ocurre porque las velocidades articulares son cero en estos instantes, condición impuesta por la trayectoria polinómica de quinto orden para garantizar un movimiento suave y sin discontinuidades.

A partir de estos resultados puede estimarse que el sistema de accionamiento deberá ser capaz de suministrar al menos 5 W de potencia mecánica durante las condiciones más exigentes de operación.
